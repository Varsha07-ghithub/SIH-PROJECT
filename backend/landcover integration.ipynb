```python
# ============================================================
# NASA FIRMS → ONE DATASET → DYNAMIC WORLD LAND-COVER
# Dataset: SV-C2-794014
# ============================================================

import os
import zipfile
import pandas as pd
import ee

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

ZIP_PATH = "/content/drive/MyDrive/datasets/DL_FIRE_SV-C2_794014.zip"

EXTRACT_DIR = "/content/drive/MyDrive/datasets/DL_FIRE_SV-C2_794014"

CSV_PATH = os.path.join(
    EXTRACT_DIR,
    "fire_archive_SV-C2_794014.csv"
)

OUTPUT_CSV = "/content/firms_sv_dynamic_world_100.csv"

# YOUR REAL GOOGLE CLOUD PROJECT ID
PROJECT_ID = "rosy-phalanx-507408-c2"


# ------------------------------------------------------------
# 2. EARTH ENGINE AUTHENTICATION
# ------------------------------------------------------------

print("🔐 Initializing Google Earth Engine...")

try:
    ee.Initialize(project=PROJECT_ID)
    print("✅ Earth Engine already initialized")

except Exception:
    print("⚠️ Authentication required...")

    ee.Authenticate()

    ee.Initialize(project=PROJECT_ID)

    print("✅ Earth Engine initialized")


print("🌍 Earth Engine Project:", PROJECT_ID)


# ------------------------------------------------------------
# 3. TEST EARTH ENGINE
# ------------------------------------------------------------

test = ee.Number(10).add(20).getInfo()

print("🧪 Earth Engine test:", test)

if test != 30:
    raise RuntimeError("❌ Earth Engine test failed")

print("🎉 Earth Engine is working!")


# ------------------------------------------------------------
# 4. EXTRACT FIRMS ZIP
# ------------------------------------------------------------

if not os.path.exists(CSV_PATH):

    print("\n📦 Extracting FIRMS ZIP...")

    os.makedirs(EXTRACT_DIR, exist_ok=True)

    with zipfile.ZipFile(ZIP_PATH, "r") as z:
        z.extractall(EXTRACT_DIR)

    print("✅ ZIP extracted")

else:

    print("\n✅ FIRMS CSV already extracted")


# ------------------------------------------------------------
# 5. LOAD ONLY ONE DATASET
# ------------------------------------------------------------

print("\n📂 Loading SV-C2 FIRMS dataset...")

df = pd.read_csv(CSV_PATH)

print("🔥 Total FIRMS records:", len(df))


# ------------------------------------------------------------
# 6. CLEAN COORDINATES
# ------------------------------------------------------------

df["latitude"] = pd.to_numeric(
    df["latitude"],
    errors="coerce"
)

df["longitude"] = pd.to_numeric(
    df["longitude"],
    errors="coerce"
)

df = df.dropna(
    subset=["latitude", "longitude"]
)

print("✅ Coordinates cleaned")


# ------------------------------------------------------------
# 7. INDIA BOUNDING BOX
# ------------------------------------------------------------

df = df[
    (df["latitude"] >= 6) &
    (df["latitude"] <= 37) &
    (df["longitude"] >= 68) &
    (df["longitude"] <= 98)
].copy()

print("🇮🇳 Records inside India:", len(df))


# ------------------------------------------------------------
# 8. SELECT 100 FIRMS POINTS FOR TESTING
# ------------------------------------------------------------

test_df = df.head(100).copy()

print("🔥 Points selected for testing:", len(test_df))


# ------------------------------------------------------------
# 9. CREATE EARTH ENGINE POINTS
# ------------------------------------------------------------

features = []

for idx, row in test_df.iterrows():

    point = ee.Geometry.Point([
        float(row["longitude"]),
        float(row["latitude"])
    ])

    properties = {
        "event_id": int(idx),
        "latitude": float(row["latitude"]),
        "longitude": float(row["longitude"])
    }

    # FIRMS attributes
    for col in [
        "acq_date",
        "acq_time",
        "satellite",
        "instrument",
        "confidence",
        "frp",
        "brightness",
        "bright_ti4"
    ]:

        if col in row.index:

            value = row[col]

            if pd.notna(value):

                # Keep numeric values numeric where possible
                if col in ["frp", "brightness", "bright_ti4"]:

                    try:
                        properties[col] = float(value)

                    except:
                        properties[col] = str(value)

                else:

                    properties[col] = str(value)

    features.append(
        ee.Feature(
            point,
            properties
        )
    )


firms_points = ee.FeatureCollection(features)

print("✅ FIRMS EE points created")


# ------------------------------------------------------------
# 10. DYNAMIC WORLD
# ------------------------------------------------------------

print("\n🛰️ Loading Dynamic World...")

region = firms_points.geometry().bounds()

dynamic_world = (
    ee.ImageCollection(
        "GOOGLE/DYNAMICWORLD/V1"
    )
    .filterBounds(region)
    .filterDate(
        "2024-01-01",
        "2025-01-01"
    )
)

dw_count = dynamic_world.size().getInfo()

print(
    "🛰️ Dynamic World images:",
    dw_count
)

if dw_count == 0:

    raise RuntimeError(
        "❌ No Dynamic World images found for the selected FIRMS points."
    )


# ------------------------------------------------------------
# 11. MOST COMMON LAND-COVER CLASS
# ------------------------------------------------------------

print("\n🌱 Creating 2024 land-cover image...")

landcover_image = (
    dynamic_world
    .select("label")
    .mode()
)

print("✅ 2024 Dynamic World mode image created")


# ------------------------------------------------------------
# 12. SAMPLE LAND COVER AT FIRMS POINTS
# ------------------------------------------------------------

print("\n🌱 Extracting land-cover classes...")

sampled_points = landcover_image.sampleRegions(
    collection=firms_points,
    scale=10,
    geometries=True
)

results = sampled_points.getInfo()

print("✅ Land-cover extraction completed")


# ------------------------------------------------------------
# 13. LAND-COVER CLASS NAMES
# ------------------------------------------------------------

LANDCOVER_CLASSES = {

    0: "Water",

    1: "Trees / Forest",

    2: "Grass",

    3: "Flooded Vegetation",

    4: "Agriculture / Crops",

    5: "Shrub and Scrub",

    6: "Built Area",

    7: "Bare Ground",

    8: "Snow and Ice"
}


# ------------------------------------------------------------
# 14. CREATE RESULT DATAFRAME
# ------------------------------------------------------------

output_rows = []

for feature in results["features"]:

    props = feature["properties"]

    landcover_id = props.get("label")

    try:

        landcover_id = int(landcover_id)

        landcover_name = LANDCOVER_CLASSES.get(
            landcover_id,
            "Unknown"
        )

    except:

        landcover_id = -1
        landcover_name = "Unknown"


    props["landcover_id"] = landcover_id

    props["landcover"] = landcover_name

    output_rows.append(props)


landcover_results_df = pd.DataFrame(
    output_rows
)


# ------------------------------------------------------------
# 15. SAVE RESULT
# ------------------------------------------------------------

landcover_results_df.to_csv(
    OUTPUT_CSV,
    index=False
)


# ------------------------------------------------------------
# 16. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("✅ EXTRACTION COMPLETE")
print("=" * 70)

print(
    "📊 Total FIRMS records:",
    len(df)
)

print(
    "🔥 FIRMS points processed:",
    len(landcover_results_df)
)

print(
    "💾 Output file:",
    OUTPUT_CSV
)

print("\n🌱 Land-cover distribution:")

print(
    landcover_results_df[
        "landcover"
    ].value_counts()
)

print("\n📋 First 20 results:")

display(
    landcover_results_df.head(20)
)
```

```python
# ================================================================
# FAST METHOD
# 1.2 MILLION FIRMS POINTS
# DYNAMIC WORLD LAND COVER
# EARTH ENGINE BATCH EXPORT
# ================================================================

import ee
import pandas as pd
import numpy as np

# ------------------------------------------------
# EARTH ENGINE
# ------------------------------------------------

PROJECT_ID = "rosy-phalanx-507408-c2"

try:
    ee.Initialize(project=PROJECT_ID)
except:
    ee.Authenticate()
    ee.Initialize(project=PROJECT_ID)

print("✅ Earth Engine ready")


# ------------------------------------------------
# LOAD YOUR CLEAN FIRMS DATA
# ------------------------------------------------

FIRMS_FILE = (
    "/content/drive/MyDrive/datasets/"
    "FIRMS_SV_2YEAR_DYNAMIC_WORLD/"
    "FIRMS_SV_ALL_POINTS_2024_2025.parquet"
)

df = pd.read_parquet(FIRMS_FILE)

print(
    "🔥 FIRMS points:",
    len(df)
)


# ------------------------------------------------
# IMPORTANT:
# Keep ONLY columns needed for the export
# ------------------------------------------------

keep_cols = [
    "event_id",
    "latitude",
    "longitude",
    "acq_date",
    "acq_time",
    "satellite",
    "instrument",
    "brightness",
    "confidence",
    "frp",
    "risk_score",
    "risk_level"
]

keep_cols = [
    c for c in keep_cols
    if c in df.columns
]

df = df[keep_cols].copy()


# ------------------------------------------------
# CONVERT DATE TO STRING
# ------------------------------------------------

if "acq_date" in df.columns:

    df["acq_date"] = (
        pd.to_datetime(
            df["acq_date"],
            errors="coerce"
        )
        .dt.strftime("%Y-%m-%d")
        .fillna("")
    )


# ------------------------------------------------
# CREATE EARTH ENGINE FEATURES
# ------------------------------------------------

print(
    "\n📍 Creating Earth Engine points..."
)

features = []

for row in df.itertuples(index=False):

    properties = {}

    for col in keep_cols:

        if col in [
            "latitude",
            "longitude"
        ]:
            continue

        value = getattr(
            row,
            col
        )

        if pd.isna(value):
            continue

        # Convert numpy types
        if isinstance(
            value,
            (np.integer,)
        ):
            value = int(value)

        elif isinstance(
            value,
            (np.floating,)
        ):
            value = float(value)

        else:
            value = str(value)

        properties[col] = value


    point = ee.Geometry.Point([
        float(row.longitude),
        float(row.latitude)
    ])

    features.append(
        ee.Feature(
            point,
            properties
        )
    )


firms_fc = ee.FeatureCollection(
    features
)

print(
    "✅ Earth Engine FeatureCollection created"
)


# ------------------------------------------------
# DYNAMIC WORLD
# ------------------------------------------------

print(
    "\n🛰️ Loading Dynamic World..."
)

dw = (
    ee.ImageCollection(
        "GOOGLE/DYNAMICWORLD/V1"
    )
    .filterDate(
        "2024-01-01",
        "2026-01-01"
    )
    .filterBounds(
        ee.Geometry.Rectangle(
            [68, 6, 98, 37]
        )
    )
)

print(
    "🛰️ Dynamic World images:",
    dw.size().getInfo()
)


# ------------------------------------------------
# TWO-YEAR MODE LAND COVER
# ------------------------------------------------

print(
    "\n🌱 Creating land-cover composite..."
)

dw_mode = (
    dw
    .select("label")
    .mode()
    .rename("landcover_id")
)


# ------------------------------------------------
# SAMPLE ALL FIRMS POINTS
# ------------------------------------------------

print(
    "\n🔥 Sampling land cover at ALL FIRMS points..."
)

result = dw_mode.sampleRegions(
    collection=firms_fc,
    properties=keep_cols,
    scale=10,
    geometries=True
)

print(
    "✅ Sampling operation created"
)


# ------------------------------------------------
# EXPORT TO GOOGLE DRIVE
# ------------------------------------------------

print(
    "\n🚀 Starting Earth Engine export..."
)

task = ee.batch.Export.table.toDrive(
    collection=result,

    description=(
        "FIRMS_SV_2024_2025_ALL_DYNAMIC_WORLD"
    ),

    folder=(
        "FIRMS_SV_2YEAR_DYNAMIC_WORLD"
    ),

    fileNamePrefix=(
        "FIRMS_SV_2024_2025_ALL_DYNAMIC_WORLD"
    ),

    fileFormat="CSV",

    selectors=(
        keep_cols +
        [
            "landcover_id",
            ".geo"
        ]
    ),

    priority=9999
)

task.start()

print("\n" + "=" * 70)
print("🚀 EXPORT STARTED")
print("=" * 70)

print(
    "🔥 FIRMS points:",
    f"{len(df):,}"
)

print(
    "🌱 Dynamic World:",
    "10 m"
)

print(
    "📅 Period:",
    "2024-01-01 → 2025-12-31"
)

print(
    "📁 Drive folder:",
    "FIRMS_SV_2YEAR_DYNAMIC_WORLD"
)

print(
    "\n⏳ Earth Engine is processing the job in the background."
)

print(
    "You do NOT need to keep running getInfo()."
)
```

```python
# ============================================================
# STEP 19A — FIRMS + DYNAMIC WORLD TEST
# Using project: rosy-phalanx-507408-c2
# ============================================================

import ee
import pandas as pd
import numpy as np

PROJECT_ID = "rosy-phalanx-507408-c2"

# ------------------------------------------------------------
# 1. Initialize Earth Engine
# ------------------------------------------------------------
ee.Initialize(project=PROJECT_ID)

print("✅ Earth Engine initialized")
print("Project:", PROJECT_ID)


# ------------------------------------------------------------
# 2. FIRMS CSV paths
# ------------------------------------------------------------
FIRMS_1 = "/content/drive/MyDrive/datasets/DL_FIRE_SV-C2_794014/fire_archive_SV-C2_794014.csv"

FIRMS_2 = "/content/drive/MyDrive/datasets/DL_FIRE_J2V-C2_794013/fire_nrt_J2V-C2_794013.csv"


# ------------------------------------------------------------
# 3. Load both FIRMS datasets
# ------------------------------------------------------------
print("\n📂 Loading FIRMS datasets...")

df1 = pd.read_csv(FIRMS_1)
df2 = pd.read_csv(FIRMS_2)

print("Dataset 1:", len(df1))
print("Dataset 2:", len(df2))


# ------------------------------------------------------------
# 4. Combine datasets
# ------------------------------------------------------------
df_firms = pd.concat(
    [df1, df2],
    ignore_index=True
)

print("Combined FIRMS records:", len(df_firms))


# ------------------------------------------------------------
# 5. Clean coordinates
# ------------------------------------------------------------
df_firms["latitude"] = pd.to_numeric(
    df_firms["latitude"],
    errors="coerce"
)

df_firms["longitude"] = pd.to_numeric(
    df_firms["longitude"],
    errors="coerce"
)

df_firms = df_firms.dropna(
    subset=["latitude", "longitude"]
)


# ------------------------------------------------------------
# 6. India bounding box
# ------------------------------------------------------------
df_firms = df_firms[
    (df_firms["latitude"] >= 6) &
    (df_firms["latitude"] <= 37) &
    (df_firms["longitude"] >= 68) &
    (df_firms["longitude"] <= 98)
].copy()

print("India FIRMS records:", len(df_firms))


# ------------------------------------------------------------
# 7. Take only 1,000 points for TEST
# ------------------------------------------------------------
TEST_POINTS = 1000

sample_df = df_firms.sample(
    n=min(TEST_POINTS, len(df_firms)),
    random_state=42
).reset_index(drop=True)

print("Test points:", len(sample_df))


# ------------------------------------------------------------
# 8. Create Earth Engine FeatureCollection
# ------------------------------------------------------------
features = []

for i, row in sample_df.iterrows():

    properties = {
        "event_id": int(i),
        "latitude": float(row["latitude"]),
        "longitude": float(row["longitude"])
    }

    # Add FIRMS fields if available
    for col in ["frp", "brightness", "confidence", "satellite", "acq_date"]:
        if col in row.index:

            value = row[col]

            if pd.notna(value):

                try:
                    if col in ["frp", "brightness"]:
                        properties[col] = float(value)

                    else:
                        properties[col] = str(value)

                except:
                    properties[col] = str(value)

    point = ee.Geometry.Point([
        float(row["longitude"]),
        float(row["latitude"])
    ])

    features.append(
        ee.Feature(point, properties)
    )


firms_points = ee.FeatureCollection(features)

print("✅ Earth Engine FIRMS FeatureCollection created")


# ------------------------------------------------------------
# 9. Get region
# ------------------------------------------------------------
region = firms_points.geometry().bounds()

print("✅ FIRMS region created")


# ------------------------------------------------------------
# 10. Load Dynamic World
# ------------------------------------------------------------
print("\n🌍 Loading Dynamic World...")

dynamic_world = (
    ee.ImageCollection("GOOGLE/DYNAMICWORLD/V1")
    .filterBounds(region)
    .filterDate(
        "2024-01-01",
        "2025-01-01"
    )
)

dw_count = dynamic_world.size().getInfo()

print("Dynamic World images:", dw_count)


# ------------------------------------------------------------
# 11. Create land-cover composite
# ------------------------------------------------------------
print("\n🛰️ Creating Dynamic World land-cover composite...")

landcover = (
    dynamic_world
    .select("label")
    .mode()
)

print("✅ Land-cover composite created")


# ------------------------------------------------------------
# 12. Sample land cover at FIRMS points
# ------------------------------------------------------------
print("\n📍 Sampling land cover at FIRMS points...")

sampled = landcover.sampleRegions(
    collection=firms_points,
    scale=10,
    geometries=True
)

result = sampled.limit(1000).getInfo()

print("✅ Sampling successful")


# ------------------------------------------------------------
# 13. Display sample results
# ------------------------------------------------------------
rows = []

for feature in result["features"]:

    p = feature["properties"]

    rows.append(p)


result_df = pd.DataFrame(rows)

print("\n" + "=" * 70)
print("🎉 FIRMS + DYNAMIC WORLD TEST SUCCESSFUL")
print("=" * 70)

print("Total tested:", len(result_df))

if "label" in result_df.columns:

    print("\nDynamic World classes found:")
    print(result_df["label"].value_counts().sort_index())


print("\nSample:")
display(result_df.head(10))
```

```python
# ============================================================
# ALL 2.33 MILLION FIRMS POINTS - FOLIUM GOOGLE SATELLITE MAP
# Scalable visualization using spatial rasterization
# ============================================================

import numpy as np
import pandas as pd
import folium
from folium.raster_layers import ImageOverlay
from PIL import Image
from IPython.display import display
import matplotlib.pyplot as plt

print("🔥 Preparing ALL FIRMS points...")

# ------------------------------------------------------------
# 1. Make sure latitude / longitude are numeric
# ------------------------------------------------------------
df = df_firms.copy()

df["latitude"] = pd.to_numeric(df["latitude"], errors="coerce")
df["longitude"] = pd.to_numeric(df["longitude"], errors="coerce")

df = df.dropna(subset=["latitude", "longitude"])

# India bounding box
df = df[
    (df["latitude"] >= 6) &
    (df["latitude"] <= 37) &
    (df["longitude"] >= 68) &
    (df["longitude"] <= 98)
]

print("Total FIRMS records:", len(df))

# ------------------------------------------------------------
# 2. Extract ALL coordinates
# ------------------------------------------------------------
lat = df["latitude"].to_numpy(dtype=np.float32)
lon = df["longitude"].to_numpy(dtype=np.float32)

print("Points used:", len(lat))

# ------------------------------------------------------------
# 3. Spatial rasterization
#
# Smaller resolution = more detail
# 0.03 degrees gives good balance for India
# ------------------------------------------------------------
GRID_SIZE = 0.03

lon_min = 68
lon_max = 98
lat_min = 6
lat_max = 37

nx = int((lon_max - lon_min) / GRID_SIZE) + 1
ny = int((lat_max - lat_min) / GRID_SIZE) + 1

print("Raster size:", nx, "x", ny)

# ------------------------------------------------------------
# 4. Convert ALL points into density grid
# ------------------------------------------------------------
density, lat_edges, lon_edges = np.histogram2d(
    lat,
    lon,
    bins=[
        np.arange(lat_min, lat_max + GRID_SIZE, GRID_SIZE),
        np.arange(lon_min, lon_max + GRID_SIZE, GRID_SIZE)
    ]
)

print("ALL 2.33M points rasterized successfully")

# ------------------------------------------------------------
# 5. Log scaling
# This allows low-density areas to remain visible
# ------------------------------------------------------------
density_log = np.log1p(density)

if density_log.max() > 0:
    density_norm = density_log / density_log.max()
else:
    density_norm = density_log

# ------------------------------------------------------------
# 6. Create transparent RGBA image
# ------------------------------------------------------------
fig = plt.figure(figsize=(16, 12), dpi=100)

plt.imshow(
    density_norm,
    origin="lower",
    extent=[
        lon_min,
        lon_max,
        lat_min,
        lat_max
    ],
    cmap="hot"
)

plt.axis("off")
plt.tight_layout(pad=0)

temp_image = "/content/firms_all_2333226_density.png"

plt.savefig(
    temp_image,
    bbox_inches="tight",
    pad_inches=0,
    transparent=True
)

plt.close(fig)

print("Density image created")

# ------------------------------------------------------------
# 7. Create Google Satellite Folium map
# ------------------------------------------------------------
m = folium.Map(
    location=[22.5, 80.0],
    zoom_start=5,
    tiles=None,
    control_scale=True
)

# Google Satellite
folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}",
    attr="Google Satellite",
    name="Google Satellite",
    overlay=False,
    control=True
).add_to(m)

# Google Hybrid
folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=y&x={x}&y={y}&z={z}",
    attr="Google Hybrid",
    name="Google Hybrid",
    overlay=False,
    control=True
).add_to(m)

# ------------------------------------------------------------
# 8. ALL FIRMS points represented in one raster layer
# ------------------------------------------------------------
ImageOverlay(
    image=temp_image,
    bounds=[
        [lat_min, lon_min],
        [lat_max, lon_max]
    ],
    opacity=0.65,
    interactive=True,
    cross_origin=False,
    zindex=2,
    name="🔥 ALL 2,333,226 FIRMS Events"
).add_to(m)

# ------------------------------------------------------------
# 9. Layer control
# ------------------------------------------------------------
folium.LayerControl(collapsed=False).add_to(m)

# ------------------------------------------------------------
# 10. Fit India
# ------------------------------------------------------------
m.fit_bounds([
    [lat_min, lon_min],
    [lat_max, lon_max]
])

print("✅ ALL 2,333,226 FIRMS EVENTS VISUALIZED")
print("🔥 No sampling")
print("🔥 No marker clustering")
print("🔥 All events contribute to the density layer")

display(m)
```

```python
# ============================================================
# DISPLAY ALL FIRMS AS RED POINTS ON FOLIUM GOOGLE SATELLITE
# ============================================================

import folium
from folium import MacroElement
from branca.element import Template

print("🔥 Preparing FIRMS points...")

# ------------------------------------------------------------
# FIRMS coordinates
# ------------------------------------------------------------
df = df_firms.copy()

df["latitude"] = pd.to_numeric(df["latitude"], errors="coerce")
df["longitude"] = pd.to_numeric(df["longitude"], errors="coerce")

df = df.dropna(subset=["latitude", "longitude"])

# India
df = df[
    (df["latitude"] >= 6) &
    (df["latitude"] <= 37) &
    (df["longitude"] >= 68) &
    (df["longitude"] <= 98)
]

print("🔥 Total FIRMS points:", len(df))

# Convert to compact coordinate array
coords = df[["latitude", "longitude"]].to_numpy(dtype="float32")

# ------------------------------------------------------------
# Create map
# ------------------------------------------------------------
m = folium.Map(
    location=[22.5, 80],
    zoom_start=5,
    tiles=None,
    prefer_canvas=True
)

# Google Satellite
folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}",
    attr="Google Satellite",
    name="Google Satellite",
    overlay=False
).add_to(m)

# Google Hybrid
folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=y&x={x}&y={y}&z={z}",
    attr="Google Hybrid",
    name="Google Hybrid",
    overlay=False
).add_to(m)

# ------------------------------------------------------------
# Create JavaScript point array
# ------------------------------------------------------------
points_js = "[" + ",".join(
    f"[{lat:.5f},{lon:.5f}]"
    for lat, lon in coords
) + "]"

# ------------------------------------------------------------
# Canvas layer JavaScript
# ------------------------------------------------------------
template = f"""
<script>
(function() {{

    var map = {{this._parent.get_name()}};

    var points = {points_js};

    console.log("FIRMS points loaded:", points.length);

    var canvas = L.DomUtil.create('canvas', 'firms-canvas');

    canvas.style.position = 'absolute';
    canvas.style.pointerEvents = 'none';
    canvas.style.zIndex = 400;

    map.getPanes().overlayPane.appendChild(canvas);

    var ctx = canvas.getContext('2d');

    function resizeCanvas() {{
        canvas.width = map.getSize().x;
        canvas.height = map.getSize().y;
        drawPoints();
    }}

    function drawPoints() {{

        ctx.clearRect(
            0,
            0,
            canvas.width,
            canvas.height
        );

        ctx.fillStyle = 'rgba(255, 40, 0, 0.75)';

        var bounds = map.getBounds();

        for (var i = 0; i < points.length; i++) {{

            var lat = points[i][0];
            var lon = points[i][1];

            // Skip points outside current map area
            if (!bounds.contains([lat, lon]))
                continue;

            var p = map.latLngToContainerPoint([lat, lon]);

            ctx.fillRect(
                p.x,
                p.y,
                2,
                2
            );
        }}
    }}

    map.on('moveend zoomend resize', function() {{
        resizeCanvas();
    }});

    resizeCanvas();

    console.log("🔥 ALL FIRMS POINTS DISPLAYED");

}})();
</script>
"""

macro = MacroElement()
macro._template = Template(template)

m.get_root().add_child(macro)

# ------------------------------------------------------------
# Layer control
# ------------------------------------------------------------
folium.LayerControl(collapsed=False).add_to(m)

# India view
m.fit_bounds([
    [6, 68],
    [37, 98]
])

print("======================================")
print("🔥 ALL FIRMS POINTS ADDED")
print("📍 POINTS:", len(df))
print("🛰️ GOOGLE SATELLITE ENABLED")
print("======================================")

display(m)
```

```python
# ============================================================
# 🔥 INDUSTRIAL FIRE INTELLIGENCE
# PYDECK + GOOGLE SATELLITE + DYNAMIC WORLD BUILT AREA
# ============================================================

!pip -q install pydeck

import pandas as pd
import numpy as np
import pydeck as pdk
import ee

from IPython.display import display, HTML
from google.colab import output

# ============================================================
# 1. EARTH ENGINE
# ============================================================

PROJECT_ID = "rosy-phalanx-507408-c2"

ee.Initialize(project=PROJECT_ID)

print("✅ Earth Engine initialized")

# ============================================================
# 2. FIRMS DATA
# ============================================================

print("🔥 Loading FIRMS data...")

df = df_firms.copy()

df["latitude"] = pd.to_numeric(
    df["latitude"],
    errors="coerce"
)

df["longitude"] = pd.to_numeric(
    df["longitude"],
    errors="coerce"
)

df = df.dropna(
    subset=["latitude", "longitude"]
)

# India
df = df[
    (df["latitude"] >= 6) &
    (df["latitude"] <= 37) &
    (df["longitude"] >= 68) &
    (df["longitude"] <= 98)
].copy()

print(
    f"🔥 Total FIRMS events: {len(df):,}"
)

# ============================================================
# 3. FIND COLUMNS
# ============================================================

def find_col(names):

    for c in names:

        if c in df.columns:
            return c

    return None


frp_col = find_col([
    "frp",
    "FRP"
])

brightness_col = find_col([
    "bright_ti4",
    "bright_ti5",
    "brightness"
])

confidence_col = find_col([
    "confidence",
    "CONFIDENCE"
])

date_col = find_col([
    "acq_date",
    "date"
])

time_col = find_col([
    "acq_time",
    "time"
])

satellite_col = find_col([
    "satellite",
    "Satellite"
])

# ============================================================
# 4. STANDARDIZE
# ============================================================

if frp_col:

    df["FRP"] = pd.to_numeric(
        df[frp_col],
        errors="coerce"
    ).fillna(0)

else:

    df["FRP"] = 0


if brightness_col:

    df["Brightness"] = pd.to_numeric(
        df[brightness_col],
        errors="coerce"
    ).fillna(0)

else:

    df["Brightness"] = 0


if confidence_col:

    df["Confidence"] = (
        df[confidence_col]
        .astype(str)
    )

else:

    df["Confidence"] = "Unknown"


if date_col:

    df["Fire_Date"] = (
        df[date_col]
        .astype(str)
    )

else:

    df["Fire_Date"] = "Unknown"


if time_col:

    df["Fire_Time"] = (
        df[time_col]
        .astype(str)
    )

else:

    df["Fire_Time"] = "Unknown"


if satellite_col:

    df["Satellite"] = (
        df[satellite_col]
        .astype(str)
    )

else:

    df["Satellite"] = "Unknown"


# ============================================================
# 5. RISK SCORE
# ============================================================

def percentile_score(series):

    x = pd.to_numeric(
        series,
        errors="coerce"
    ).fillna(0)

    if x.nunique() <= 1:

        return pd.Series(
            0,
            index=x.index
        )

    return x.rank(
        pct=True
    ) * 100


frp_score = percentile_score(
    df["FRP"]
)

brightness_score = percentile_score(
    df["Brightness"]
)


confidence_map = {
    "l": 30,
    "n": 60,
    "h": 90,
    "low": 30,
    "nominal": 60,
    "high": 90
}


confidence_numeric = (
    df["Confidence"]
    .astype(str)
    .str.lower()
    .map(confidence_map)
)

confidence_numeric = (
    confidence_numeric
    .fillna(
        pd.to_numeric(
            df["Confidence"],
            errors="coerce"
        )
    )
    .fillna(0)
)


confidence_score = percentile_score(
    confidence_numeric
)


df["Risk_Score"] = (
    0.50 * frp_score +
    0.30 * brightness_score +
    0.20 * confidence_score
).round(1)


# ============================================================
# 6. RISK LEVEL
# ============================================================

def get_level(x):

    if x >= 80:
        return "CRITICAL"

    elif x >= 60:
        return "HIGH"

    elif x >= 35:
        return "MODERATE"

    return "LOW"


df["Risk_Level"] = (
    df["Risk_Score"]
    .apply(get_level)
)


# ============================================================
# 7. INTERACTIVE DATA
# ============================================================

INTERACTIVE_POINTS = min(
    100000,
    len(df)
)

map_df = df.sample(
    INTERACTIVE_POINTS,
    random_state=42
).reset_index(drop=True)

print(
    f"📍 Interactive points: "
    f"{len(map_df):,}"
)

# ============================================================
# 8. RISK COLORS
# ============================================================

def color(level):

    if level == "CRITICAL":
        return [255, 30, 30, 230]

    if level == "HIGH":
        return [255, 100, 0, 220]

    if level == "MODERATE":
        return [255, 200, 0, 210]

    return [50, 200, 100, 180]


rgba = np.array(
    [
        color(x)
        for x in map_df["Risk_Level"]
    ],
    dtype=np.uint8
)

map_df["R"] = rgba[:, 0]
map_df["G"] = rgba[:, 1]
map_df["B"] = rgba[:, 2]
map_df["A"] = rgba[:, 3]

# ============================================================
# 9. GOOGLE SATELLITE
# ============================================================

satellite_layer = pdk.Layer(

    "TileLayer",

    data=
    "https://mt1.google.com/vt/"
    "?lyrs=s&x={x}&y={y}&z={z}",

    min_zoom=2,

    max_zoom=20,

    tile_size=256
)

# ============================================================
# 10. FIRMS POINTS
# ============================================================

fire_layer = pdk.Layer(

    "ScatterplotLayer",

    data=map_df,

    get_position=
    "[longitude, latitude]",

    get_radius=
    "Math.max(100, FRP * 80)",

    radius_min_pixels=2,

    radius_max_pixels=8,

    get_fill_color=
    "[R, G, B, A]",

    pickable=True,

    auto_highlight=True,

    opacity=0.9
)

# ============================================================
# 11. MAP
# ============================================================

view = pdk.ViewState(

    latitude=22.5,

    longitude=79.0,

    zoom=4.8,

    pitch=0,

    bearing=0
)

# ============================================================
# 12. TOOLTIP
# ============================================================

tooltip = {

    "html": """

    <div style="
        width:320px;
        background:#07111f;
        color:white;
        padding:18px;
        border-radius:12px;
        font-family:Arial;
        box-shadow:0 8px 30px rgba(0,0,0,.6);
    ">

    <div style="
        font-size:20px;
        font-weight:bold;
        color:#ff453a;
        margin-bottom:12px;
    ">
    🔥 FIRMS FIRE EVENT
    </div>

    <div style="
        background:#111c2d;
        padding:10px;
        border-radius:8px;
        margin-bottom:12px;
    ">

    <div style="
        color:#94a3b8;
        font-size:11px;
    ">
    RISK LEVEL
    </div>

    <div style="
        font-size:21px;
        font-weight:bold;
        color:#ff453a;
    ">
    {Risk_Level}
    </div>

    <div>
    Risk Score: {Risk_Score}
    </div>

    </div>

    <b>📍 Latitude:</b> {latitude}<br>
    <b>📍 Longitude:</b> {longitude}<br><br>

    <b>🔥 FRP:</b> {FRP} MW<br>
    <b>🌡 Brightness:</b> {Brightness} K<br>
    <b>📡 Confidence:</b> {Confidence}<br><br>

    <b>📅 Date:</b> {Fire_Date}<br>
    <b>⏰ Time:</b> {Fire_Time}<br>
    <b>🛰 Satellite:</b> {Satellite}

    </div>

    """,

    "style": {
        "backgroundColor": "transparent"
    }
}

# ============================================================
# 13. DASHBOARD HEADER
# ============================================================

total = len(df)

critical = int(
    (df["Risk_Level"] == "CRITICAL").sum()
)

high = int(
    (df["Risk_Level"] == "HIGH").sum()
)

moderate = int(
    (df["Risk_Level"] == "MODERATE").sum()
)

low = int(
    (df["Risk_Level"] == "LOW").sum()
)


display(
    HTML(
        f"""

        <div style="
        background:#07111f;
        color:white;
        padding:22px;
        font-family:Arial;
        border-radius:10px;
        ">

        <div style="
        font-size:30px;
        font-weight:bold;
        ">

        🔥 INDUSTRIAL FIRE INTELLIGENCE

        </div>

        <div style="
        color:#94a3b8;
        margin-top:5px;
        ">

        NASA FIRMS • Satellite Thermal Monitoring • India

        </div>


        <div style="
        display:grid;
        grid-template-columns:
        repeat(5,1fr);
        gap:12px;
        margin-top:20px;
        ">


        <div style="
        background:#151f32;
        padding:15px;
        border-radius:8px;
        ">

        <small>TOTAL EVENTS</small>

        <h2>
        {total:,}
        </h2>

        </div>


        <div style="
        background:#151f32;
        padding:15px;
        border-radius:8px;
        ">

        <small>🔴 CRITICAL</small>

        <h2 style="color:#ff3030">
        {critical:,}
        </h2>

        </div>


        <div style="
        background:#151f32;
        padding:15px;
        border-radius:8px;
        ">

        <small>🟠 HIGH</small>

        <h2 style="color:#ff8500">
        {high:,}
        </h2>

        </div>


        <div style="
        background:#151f32;
        padding:15px;
        border-radius:8px;
        ">

        <small>🟡 MODERATE</small>

        <h2 style="color:#ffd000">
        {moderate:,}
        </h2>

        </div>


        <div style="
        background:#151f32;
        padding:15px;
        border-radius:8px;
        ">

        <small>🟢 LOW</small>

        <h2 style="color:#45e080">
        {low:,}
        </h2>

        </div>

        </div>

        </div>

        """
    )
)

# ============================================================
# 14. CREATE MAP
# ============================================================

deck = pdk.Deck(

    layers=[
        satellite_layer,
        fire_layer
    ],

    initial_view_state=view,

    tooltip=tooltip,

    map_style=None,

    parameters={
        "depthTest": False
    }
)

# ============================================================
# 15. DISPLAY
# ============================================================

print()
print("==============================================")
print("🔥 INDUSTRIAL FIRE MAP READY")
print("==============================================")
print(f"🔥 Total events      : {total:,}")
print(f"📍 Interactive points: {len(map_df):,}")
print("🛰 Google Satellite  : ON")
print("🖱 Click a fire point")
print("==============================================")

deck
```

```python
# ============================================================
# 🔥 ALL FIRMS POINTS + RISK COLOURS + GOOGLE SATELLITE
# ============================================================

!pip -q install pydeck

import pandas as pd
import numpy as np
import pydeck as pdk
import os

from IPython.display import display, HTML

# ============================================================
# 1. FIRMS FILES
# ============================================================

FILE1 = "/content/drive/MyDrive/datasets/DL_FIRE_SV-C2_794014/fire_archive_SV-C2_794014.csv"

FILE2 = "/content/drive/MyDrive/datasets/DL_FIRE_J2V-C2_794013/fire_nrt_J2V-C2_794013.csv"

print("🔥 Loading FIRMS files...")

df1 = pd.read_csv(FILE1)
df2 = pd.read_csv(FILE2)

print(f"Dataset 1 : {len(df1):,}")
print(f"Dataset 2 : {len(df2):,}")

# ============================================================
# 2. COMBINE
# ============================================================

df = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(f"🔥 Combined FIRMS : {len(df):,}")

# ============================================================
# 3. CLEAN COORDINATES
# ============================================================

df["latitude"] = pd.to_numeric(
    df["latitude"],
    errors="coerce"
)

df["longitude"] = pd.to_numeric(
    df["longitude"],
    errors="coerce"
)

df = df.dropna(
    subset=["latitude", "longitude"]
)

# India bounding box
df = df[
    (df["latitude"] >= 6) &
    (df["latitude"] <= 37) &
    (df["longitude"] >= 68) &
    (df["longitude"] <= 98)
].copy()

print(
    f"🇮🇳 India FIRMS events : {len(df):,}"
)

# ============================================================
# 4. FIND FIRMS COLUMNS
# ============================================================

def find_column(possible):

    for c in possible:

        if c in df.columns:
            return c

    return None


frp_col = find_column([
    "frp",
    "FRP"
])

brightness_col = find_column([
    "bright_ti4",
    "bright_ti5",
    "brightness"
])

confidence_col = find_column([
    "confidence",
    "CONFIDENCE"
])

date_col = find_column([
    "acq_date",
    "date"
])

time_col = find_column([
    "acq_time",
    "time"
])

satellite_col = find_column([
    "satellite",
    "Satellite"
])

print()
print("Detected columns:")
print("FRP        :", frp_col)
print("Brightness :", brightness_col)
print("Confidence :", confidence_col)
print("Date       :", date_col)
print("Time       :", time_col)
print("Satellite  :", satellite_col)

# ============================================================
# 5. STANDARDIZE FIRMS DATA
# ============================================================

if frp_col:

    df["FRP"] = pd.to_numeric(
        df[frp_col],
        errors="coerce"
    ).fillna(0)

else:

    df["FRP"] = 0.0


if brightness_col:

    df["Brightness"] = pd.to_numeric(
        df[brightness_col],
        errors="coerce"
    ).fillna(0)

else:

    df["Brightness"] = 0.0


if confidence_col:

    df["Confidence"] = (
        df[confidence_col]
        .astype(str)
    )

else:

    df["Confidence"] = "Unknown"


if date_col:

    df["Fire_Date"] = (
        df[date_col]
        .astype(str)
    )

else:

    df["Fire_Date"] = "Unknown"


if time_col:

    df["Fire_Time"] = (
        df[time_col]
        .astype(str)
    )

else:

    df["Fire_Time"] = "Unknown"


if satellite_col:

    df["Satellite"] = (
        df[satellite_col]
        .astype(str)
    )

else:

    df["Satellite"] = "Unknown"


# ============================================================
# 6. RISK SCORE
# ============================================================

print()
print("⚠️ Calculating risk score...")

def percentile_score(series):

    x = pd.to_numeric(
        series,
        errors="coerce"
    ).fillna(0)

    if x.nunique() <= 1:

        return pd.Series(
            0.0,
            index=x.index
        )

    return (
        x.rank(pct=True) * 100
    )


# FRP = 50%
frp_score = percentile_score(
    df["FRP"]
)

# Brightness = 30%
brightness_score = percentile_score(
    df["Brightness"]
)

# Confidence = 20%
confidence_map = {
    "l": 30,
    "n": 60,
    "h": 90,
    "low": 30,
    "nominal": 60,
    "high": 90
}

confidence_numeric = (
    df["Confidence"]
    .astype(str)
    .str.lower()
    .map(confidence_map)
)

confidence_numeric = (
    confidence_numeric
    .fillna(
        pd.to_numeric(
            df["Confidence"],
            errors="coerce"
        )
    )
    .fillna(0)
)

confidence_score = percentile_score(
    confidence_numeric
)

# FINAL RISK
df["Risk_Score"] = (
    0.50 * frp_score +
    0.30 * brightness_score +
    0.20 * confidence_score
).round(1)


def risk_level(score):

    if score >= 80:
        return "CRITICAL"

    elif score >= 60:
        return "HIGH"

    elif score >= 35:
        return "MODERATE"

    else:
        return "LOW"


df["Risk_Level"] = (
    df["Risk_Score"]
    .apply(risk_level)
)

# ============================================================
# 7. RISK COLORS
# ============================================================

def get_color(level):

    if level == "CRITICAL":
        return [255, 0, 0, 230]

    elif level == "HIGH":
        return [255, 90, 0, 220]

    elif level == "MODERATE":
        return [255, 200, 0, 210]

    else:
        return [40, 200, 100, 190]


colors = np.array([
    get_color(x)
    for x in df["Risk_Level"]
], dtype=np.uint8)

df["R"] = colors[:, 0]
df["G"] = colors[:, 1]
df["B"] = colors[:, 2]
df["A"] = colors[:, 3]

# ============================================================
# 8. MAP DATA
# ============================================================

map_df = df[
    [
        "latitude",
        "longitude",
        "FRP",
        "Brightness",
        "Confidence",
        "Fire_Date",
        "Fire_Time",
        "Satellite",
        "Risk_Score",
        "Risk_Level",
        "R",
        "G",
        "B",
        "A"
    ]
].copy()

# Memory optimization
map_df["latitude"] = (
    map_df["latitude"]
    .astype("float32")
)

map_df["longitude"] = (
    map_df["longitude"]
    .astype("float32")
)

map_df["FRP"] = (
    map_df["FRP"]
    .astype("float32")
)

map_df["Brightness"] = (
    map_df["Brightness"]
    .astype("float32")
)

map_df["Risk_Score"] = (
    map_df["Risk_Score"]
    .astype("float32")
)

print()
print("==============================================")
print("🔥 ALL FIRMS POINTS READY")
print("==============================================")
print(f"🔥 TOTAL POINTS : {len(map_df):,}")
print("🔴 CRITICAL     :",
      (df["Risk_Level"] == "CRITICAL").sum())
print("🟠 HIGH         :",
      (df["Risk_Level"] == "HIGH").sum())
print("🟡 MODERATE     :",
      (df["Risk_Level"] == "MODERATE").sum())
print("🟢 LOW          :",
      (df["Risk_Level"] == "LOW").sum())
print("==============================================")

# ============================================================
# 9. GOOGLE SATELLITE
# ============================================================

satellite_layer = pdk.Layer(

    "TileLayer",

    data=
    "https://mt1.google.com/vt/"
    "?lyrs=s&x={x}&y={y}&z={z}",

    min_zoom=2,

    max_zoom=20,

    tile_size=256
)

# ============================================================
# 10. ALL 2.33 MILLION FIRE POINTS
# ============================================================

fire_layer = pdk.Layer(

    "ScatterplotLayer",

    data=map_df,

    get_position=[
        "longitude",
        "latitude"
    ],

    get_radius=120,

    radius_min_pixels=1,

    radius_max_pixels=5,

    get_fill_color=[
        "R",
        "G",
        "B",
        "A"
    ],

    pickable=True,

    auto_highlight=True,

    stroked=False,

    filled=True,

    opacity=0.8
)

# ============================================================
# 11. MAP VIEW
# ============================================================

view = pdk.ViewState(

    latitude=22.5,

    longitude=79.0,

    zoom=4.7,

    pitch=0,

    bearing=0
)

# ============================================================
# 12. TOOLTIP
# ============================================================

tooltip = {

    "html": """

    <div style="
        width:330px;
        background:#07111f;
        color:white;
        padding:18px;
        border-radius:14px;
        font-family:Arial;
        box-shadow:0 8px 30px rgba(0,0,0,.7);
    ">

    <div style="
        color:#ff3b30;
        font-size:21px;
        font-weight:bold;
        margin-bottom:12px;
    ">
        🔥 FIRMS FIRE
    </div>

    <div style="
        background:#121e30;
        padding:10px;
        border-radius:8px;
        margin-bottom:12px;
    ">

        <div style="
            color:#94a3b8;
            font-size:11px;
        ">
            RISK LEVEL
        </div>

        <div style="
            font-size:22px;
            font-weight:bold;
        ">
            {Risk_Level}
        </div>

        <div>
            Risk Score: <b>{Risk_Score}</b>
        </div>

    </div>

    <b>📍 Latitude:</b> {latitude}<br>
    <b>📍 Longitude:</b> {longitude}<br><br>

    <b>🔥 FRP:</b> {FRP} MW<br>
    <b>🌡 Brightness:</b> {Brightness} K<br>
    <b>📡 Confidence:</b> {Confidence}<br>
    <b>🛰 Satellite:</b> {Satellite}<br>
    <b>📅 Date:</b> {Fire_Date}<br>
    <b>⏰ Time:</b> {Fire_Time}

    <hr style="
        border:0;
        border-top:1px solid #334155;
        margin:12px 0;
    ">

    <div style="
        color:#ff4dff;
        font-size:16px;
        font-weight:bold;
    ">
        🌱 LAND COVER
    </div>

    <div style="
        color:#94a3b8;
        margin-top:5px;
    ">
        Click this fire for Dynamic World
        land-cover analysis.
    </div>

    </div>

    """,

    "style": {
        "backgroundColor": "transparent"
    }
}

# ============================================================
# 13. DASHBOARD HEADER
# ============================================================

total = len(df)

critical = int(
    (df["Risk_Level"] == "CRITICAL").sum()
)

high = int(
    (df["Risk_Level"] == "HIGH").sum()
)

moderate = int(
    (df["Risk_Level"] == "MODERATE").sum()
)

low = int(
    (df["Risk_Level"] == "LOW").sum()
)

display(
    HTML(
        f"""

        <div style="
            background:#07111f;
            color:white;
            padding:22px;
            border-radius:12px;
            font-family:Arial;
            margin-bottom:10px;
        ">

        <div style="
            font-size:30px;
            font-weight:bold;
        ">
            🔥 INDUSTRIAL FIRE INTELLIGENCE
        </div>

        <div style="
            color:#94a3b8;
            margin-top:5px;
        ">
            NASA FIRMS • Risk Analysis • Land Cover • India
        </div>

        <div style="
            display:grid;
            grid-template-columns:
            repeat(5,1fr);
            gap:12px;
            margin-top:20px;
        ">

        <div style="
            background:#141f32;
            padding:14px;
            border-radius:8px;
        ">
            <small>TOTAL EVENTS</small>
            <div style="font-size:22px;font-weight:bold;">
                {total:,}
            </div>
        </div>

        <div style="
            background:#141f32;
            padding:14px;
            border-radius:8px;
            border-left:4px solid #ff3030;
        ">
            <small>🔴 CRITICAL</small>
            <div style="
                font-size:22px;
                font-weight:bold;
                color:#ff3030;
            ">
                {critical:,}
            </div>
        </div>

        <div style="
            background:#141f32;
            padding:14px;
            border-radius:8px;
            border-left:4px solid #ff8500;
        ">
            <small>🟠 HIGH</small>
            <div style="
                font-size:22px;
                font-weight:bold;
                color:#ff8500;
            ">
                {high:,}
            </div>
        </div>

        <div style="
            background:#141f32;
            padding:14px;
            border-radius:8px;
            border-left:4px solid #ffd000;
        ">
            <small>🟡 MODERATE</small>
            <div style="
                font-size:22px;
                font-weight:bold;
                color:#ffd000;
            ">
                {moderate:,}
            </div>
        </div>

        <div style="
            background:#141f32;
            padding:14px;
            border-radius:8px;
            border-left:4px solid #45df80;
        ">
            <small>🟢 LOW</small>
            <div style="
                font-size:22px;
                font-weight:bold;
                color:#45df80;
            ">
                {low:,}
            </div>
        </div>

        </div>

        </div>

        """
    )
)

# ============================================================
# 14. CREATE DECK
# ============================================================

deck = pdk.Deck(

    layers=[
        satellite_layer,
        fire_layer
    ],

    initial_view_state=view,

    tooltip=tooltip,

    map_style=None,

    parameters={
        "depthTest": False
    }
)

# ============================================================
# 15. DISPLAY
# ============================================================

print()
print("==============================================")
print("🔥 INDUSTRIAL FIRE MAP READY")
print("==============================================")
print(f"🔥 ALL FIRMS POINTS : {len(map_df):,}")
print("🛰 GOOGLE SATELLITE : ON")
print("🔴 CRITICAL         : RED")
print("🟠 HIGH             : ORANGE")
print("🟡 MODERATE         : YELLOW")
print("🟢 LOW              : GREEN")
print("==============================================")

deck
```
